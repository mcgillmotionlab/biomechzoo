This notebook demonstrates how to use the BiomechZoo library to extract event values from biomechanical data. The analysis includes data preprocessing, event detection, and extraction of event values for statistical analysis.

```python
# imports
import os
from biomechzoo.biomechzoo import BiomechZoo
from biomechzoo.statistics.eventval import eventval
from biomechzoo.utils.find_repo_root import find_repo_root
```

```python
# get raw data folder
project_root = find_repo_root(os.getcwd())
fld_data = os.path.join(project_root, 'data', 'sample_study', 'raw c3d files')
```

```python
# initialize biomechzoo object
bmech = BiomechZoo(fld_data, inplace=False, verbose=1)
```

```python
# Perform basic processing to showcase eventval functionality
bmech.c3d2zoo(out_folder='processing_for_eventval') # Conversion to the biomechZoo format
bmech.removechannel(ch=['RHipAngles','RKneeAngles','RAnkleAngles'], mode='keep', inplace=False)
bmech.explodechannel()
bmech.addevent(ch='RHipAngles_x', event_type='max', event_name='max')       # Find some max values
bmech.addevent(ch='RKneeAngles_x', event_type='min', event_name='min')      # Find some min values
bmech.addevent(ch='RKneeAngles_x', event_type='first', event_name='first', constant=2)     # Find FS & FO
```

```python
# Run eventval to extract events from each zoo file
dim1 = ['Straight', 'Turn']
dim2 = ['HC002D', 'HC030A', 'HC033A', 'HC040A']
local_events = ['max', 'min']
global_events = ['first']
ch = ['RHipAngles_x', 'RKneeAngles_x', 'RAnkleAngles_x', 'bad channel']
df = eventval(fld=bmech.in_folder,
              dim1=dim1,
              dim2=dim2,
              ch=ch,
              localevts=local_events,
              globalevts=global_events)
```

```python
# save event data to csv and show preview
df.to_csv('eventval_output.csv')
df.head()
```
